```python
import os
import glob
import pandas as pd
```

```python
# Parameters: adjust these to match your folder structure
DRIVE_ROOT = "/content/drive/My Drive/AI Financial Analisys/Summer Project/Vasco/Data"
INPUT_FOLDER = os.path.join(DRIVE_ROOT, "SIP")
OUTPUT_FOLDER = os.path.join(DRIVE_ROOT, "Regular Hours")
```

```python
# Number of files to process (set to 8)
max_files = 8
```

```python
# Function to process one file: filter to Regular Trading Hours (RTH)
def filter_regular_trading_hours(
    input_path,
    output_path,
    time_col=None,
    tz_source="UTC",
    tz_target="America/New_York",
    start_time="09:30",
    end_time="16:00",
):
    # Read CSV, parse datetime column or first column
    if time_col:
        df = pd.read_csv(input_path, parse_dates=[time_col])
        df.set_index(time_col, inplace=True)
    else:
        df = pd.read_csv(input_path, parse_dates=[0], index_col=0)

    # Ensure index is timezone-aware in tz_source, then convert to tz_target
    if df.index.tz is None:
        df.index = df.index.tz_localize(tz_source)
    df.index = df.index.tz_convert(tz_target)

    # Filter between regular trading hours
    df_rth = df.between_time(start_time, end_time)

    # Optionally drop timezone info for saving
    df_rth.index = df_rth.index.tz_localize(None)

    # Save to CSV
    df_rth.to_csv(output_path)
    print(f"Saved RTH data to {output_path}")
```

```python
# Get all CSV paths and limit to first `max_files`
all_files = glob.glob(os.path.join(INPUT_FOLDER, "*.csv"))
files_to_process = all_files[:max_files]

if not files_to_process:
    print("No CSV files found to process.")
else:
    print(f"Processing {len(files_to_process)} file(s)...")
    for file_path in files_to_process:
        fname = os.path.basename(file_path)
        name, ext = os.path.splitext(fname)
        out_path = os.path.join(OUTPUT_FOLDER, f"RTH_{name}_RTH{ext}")
        filter_regular_trading_hours(file_path, out_path)

print("Done.")
```
